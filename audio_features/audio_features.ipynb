```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
# Step 1: Load and preprocess the audio file
audio_file = '/content/violin_c.wav'
y, sr = librosa.load(audio_file, sr=None)  # Load audio file with its native sampling rate
```

```python
# Step 2: Feature extraction

# MFCCs
mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)  # Extract 13 MFCCs
```

```python
# Voice Spectral Energy Features
spectral_energy = np.abs(librosa.stft(y))  # STFT of the audio
voice_energy = np.sum(spectral_energy, axis=0)  # Sum the energy across all frequencies
```

```python
# Loudness (RMS Energy)
rms = librosa.feature.rms(y=y)
loudness = librosa.amplitude_to_db(rms, ref=np.max)
```

```python
# Energy
energy = np.sum(y ** 2)
```

```python
# Fundamental Frequency (Pitch)
f0, voiced_flag, voiced_probs = librosa.pyin(y, fmin=librosa.note_to_hz('C2'), fmax=librosa.note_to_hz('C7'))
```

```python
# Spectral Flux
spectral_flux = librosa.onset.onset_strength(y=y, sr=sr)
```

```python
# Spectral Centroid
spectral_centroid = librosa.feature.spectral_centroid(y=y, sr=sr)
```

```python
# Spectral Entropy
def spectral_entropy(y, sr, n_fft=2048, hop_length=512):
    S = np.abs(librosa.stft(y, n_fft=n_fft, hop_length=hop_length))
    S_norm = S / S.sum(axis=0, keepdims=True)
    return -np.sum(S_norm * np.log(S_norm + 1e-10), axis=0)

spectral_entropy = spectral_entropy(y, sr)
```

```python
# Spectral Slope
def spectral_slope(S):
    S = np.abs(S)
    freq = np.linspace(0, sr / 2, S.shape[0])
    slope = np.polyfit(freq, S, 1)[0]
    return slope

S = np.abs(librosa.stft(y))
spectral_slope = spectral_slope(S)
```

```python
# Psychoacoustic Sharpness (requires an external model or library)
# Example here is conceptual, replace with appropriate library or method.
psychoacoustic_sharpness = librosa.feature.spectral_contrast(y=y, sr=sr)
```

```python
# Spectral Harmonicity
harmonic, percussive = librosa.effects.hpss(y)
spectral_harmonicity = np.sum(harmonic ** 2) / np.sum(y ** 2)
```

```python
import scipy
# Spectral Statistical Metrics
spectral_variance = np.var(S, axis=1)
spectral_skewness = scipy.stats.skew(S, axis=1)
spectral_kurtosis = scipy.stats.kurtosis(S, axis=1)
```

```python
# Jitter and Shimmer (typically used in voice analysis, requires specific functions)
def jitter(y, sr):
    pitches, magnitudes = librosa.core.piptrack(y=y, sr=sr)
    pitch = pitches[pitches > 0]
    diff = np.diff(pitch)
    return np.mean(np.abs(diff))

def shimmer(y, sr):
    S = np.abs(librosa.stft(y))
    return np.std(S)
```

```python
jitter_value = jitter(y, sr)
shimmer_value = shimmer(y, sr)
```

```python
# Harmonic-to-Noise Ratio (HNR) calculation typically involves specific algorithms
def harmonic_to_noise_ratio(y):
    harmonic, percussive = librosa.effects.hpss(y)
    hnr = np.sum(harmonic ** 2) / np.sum(percussive ** 2)
    return hnr

hnr = harmonic_to_noise_ratio(y)
```
